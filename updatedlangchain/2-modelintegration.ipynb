```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()  # this loads your .env file

model = init_chat_model("groq:qwen/qwen3-32b")

response = model.invoke("Write me a 200 words of paragraph on Artificial Intelligence")
print(response.content)
```

```python
model.stream("Write me a 200 words of paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words of paragraph on Artificial Intelligence"):
    print(chunk.text,end="|",flush=True)
```

### Batch: Processing done in parallel

```python
response=model.batch(
    ["Why do parrots have colourful feather?"
    "How do airplanes fly?"],
    config={
        'max_concurrency':5,
    }
)
response

```
